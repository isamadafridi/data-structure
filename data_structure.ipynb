## 🧾 Python List Methods (with examples)
Screenshot 2025-04-18 104658.png


### 1. `append()`  
➡️ Adds an item to the **end** of the list.

```python
fruits = ['apple', 'banana']
fruits.append('orange')
print(fruits)  # ['apple', 'banana', 'orange']
```

---

### 2. `insert(index, item)`  
➡️ Inserts an item **at a specific position**.

```python
fruits = ['apple', 'banana']
fruits.insert(1, 'mango')
print(fruits)  # ['apple', 'mango', 'banana']
```

---

### 3. `extend(iterable)`  
➡️ Adds multiple items from another list (or any iterable).

```python
a = [1, 2]
b = [3, 4]
a.extend(b)
print(a)  # [1, 2, 3, 4]
```

---

### 4. `remove(item)`  
➡️ Removes **first occurrence** of a specific item.

```python
nums = [1, 2, 3, 2]
nums.remove(2)
print(nums)  # [1, 3, 2]
```

---

### 5. `pop(index)`  
➡️ Removes and returns item at index.  
➡️ If no index is given, removes the **last item**.

```python
nums = [10, 20, 30]
nums.pop()        # Removes 30
nums.pop(0)       # Removes 10
print(nums)       # [20]
```

---

### 6. `index(item)`  
➡️ Returns the **index of first occurrence** of item.

```python
colors = ['red', 'green', 'blue']
print(colors.index('green'))  # 1
```

---

### 7. `count(item)`  
➡️ Returns how many times an item appears.

```python
nums = [1, 2, 2, 3, 2]
print(nums.count(2))  # 3
```

---

### 8. `sort()`  
➡️ Sorts the list in-place (changes the original list).

```python
nums = [5, 3, 1, 4]
nums.sort()
print(nums)  # [1, 3, 4, 5]
```

📌 Use `sort(reverse=True)` for descending order.

---

### 9. `reverse()`  
➡️ Reverses the list **in-place**.

```python
chars = ['a', 'b', 'c']
chars.reverse()
print(chars)  # ['c', 'b', 'a']
```

---

### 10. `copy()`  
➡️ Creates a shallow **copy** of the list.

```python
a = [1, 2, 3]
b = a.copy()
b.append(4)
print(a)  # [1, 2, 3]
print(b)  # [1, 2, 3, 4]
```

---

### 11. `clear()`  
➡️ Removes **all items** from the list.

```python
items = [10, 20, 30]
items.clear()
print(items)  # []
```

**Question**: we say now List in order but this is  not in order form [12, 45, 53, 56, 67, 2, 1, 3, 4, 7, 5],why?

Ah, I see what you’re pointing out — and you’re 100% right to question it. Let’s break it down like a Python pro. 🔍🐍

---

### ✅ First, what does “**list is ordered**” mean?

It **doesn’t mean sorted**. It means:

> The order **you insert items** into the list is the **order they stay in**.

That is:
```python
a = [12, 45, 53]
# 12 is at index 0, 45 at index 1, 53 at index 2
```
That order will be preserved unless you explicitly change it.

---

### 🔄 Now your code:
```python
a = [12, 45, 53, 56, 67]
b = [2, 1, 3, 4, 7, 5]
c = a + b
print(c)
```

🟢 **Result**:
```python
[12, 45, 53, 56, 67, 2, 1, 3, 4, 7, 5]
```

This is **correct behavior**. The list `c` contains all elements of `a`, then all of `b`, in the **same order they appeared**. That's what “ordered” means in list terms.

---

### ❗️If you wanted them **sorted**, that’s a different story:

You’d need:
```python
c = a + b
c.sort()
print(c)
```

🟢 Now output would be:
```python
[1, 2, 3, 4, 5, 12, 45, 53, 56, 67, 7]
```

Oops — still not quite! Because `7` is at the end — it’s a small logic mistake. Actually, Python **sorts correctly**, but your current output might have confused you due to a typo in copying it. Correct sorting would be:

```python
[1, 2, 3, 4, 5, 7, 12, 45, 53, 56, 67]
```

Make sure all numbers are numeric types (integers or floats) and no strings mixed in.

---

### 🔑 Summary:

| Term       | Meaning in Python |
|------------|-------------------|
| Ordered    | Items stay in the order they were added |
| Changeable | You can modify, append, remove items |
| Allows duplicates | You can have repeated values |
| Sorted     | List elements are in increasing (or decreasing) order — **not automatic** |

### 🧱 Tuple in Python

A **tuple** is a built-in data type that:

### ✅ Properties:

| Property        | Description                                                                 |
|-----------------|-----------------------------------------------------------------------------|
| **Ordered**     | Items keep the order you define — just like lists                          |
| **Unchangeable**| Once created, you **can’t modify** the tuple (no append, no remove, etc.)  |
| **Allows Duplicates** | You can have the same value more than once                            |

---

### 📦 Example:

```python
coordinates = (10.5, 20.3)
```

This is a 2D coordinate — very common in math, graphics, maps, etc.

### 🔍 Let's check its properties one by one:

---

#### 1. ✅ **Ordered**

```python
coordinates = (10.5, 20.3)
print(coordinates[0])  # Output: 10.5
print(coordinates[1])  # Output: 20.3
```

- Indexing works.
- First value is always `10.5`, second is `20.3`.

---

#### 2. 🚫 **Unchangeable (Immutable)**

```python
coordinates[0] = 15.0
# ❌ TypeError: 'tuple' object does not support item assignment
```

- Tuples cannot be modified after creation.

---

#### 3. ✅ **Allows Duplicates**

```python
values = (1, 1, 2, 2, 3)
print(values)  # Output: (1, 1, 2, 2, 3)
```

- Duplicates? No problem.

---

### ✨ So why use tuples?

- **Faster** than lists (internally optimized)
- **Safer** (can’t be changed accidentally)
- Used in **functions** that return multiple values
- Great for **fixed collections** (like coordinates, RGB values, database records)

---

### 🔄 Bonus: If you need to modify a tuple?

Convert it to a list temporarily:
```python
coordinates = (10.5, 20.3)
temp = list(coordinates)
temp[0] = 15.0
coordinates = tuple(temp)
print(coordinates)  # Output: (15.0, 20.3)
```

## 🧾 Tuple Methods (Built-in)

Unlike lists, **tuples are immutable**, so they only have **two built-in methods**:

---

### 1. `count(item)`  
➡️ Returns the number of times an item appears in the tuple.

```python
numbers = (1, 2, 2, 3, 4, 2)
print(numbers.count(2))  # Output: 3
```

---

### 2. `index(item)`  
➡️ Returns the **first index** where the item appears.

```python
colors = ('red', 'green', 'blue', 'green')
print(colors.index('green'))  # Output: 1
```

---

## ⚠️ Reminder:
You **can’t** do things like `.append()`, `.remove()`, or `.sort()` with tuples, because they **cannot be changed** after creation.

---

### 🔄 Need to modify a tuple?

first tuple **convert to a list**, modify it, then **convert back to tuple**:

````python
t = (1, 2, 3)
temp = list(t)
temp.append(4)
t = tuple(temp)
print(t)
````

# **Dictionaries:**

A dictionary is an unordered, mutable collection of key-value pairs

Created using curly braces {} or the dict() function

    word_freq = {"NLP": 5, "Python": 3}

    person = {“name”: “Salman”, “age”: “may q bataon?”, “salary”: “ghalat sawal”}

Key Features of Dictionaries

-Keys are unique and immutable (e.g., strings, numbers, tuples)

-Values can be any data type

-Fast lookups by key

-Perfect for storing structured, labeled data

**Why Use Dictionaries in NLP?**

-Store word counts (term frequencies)

-Map tokens to POS tags

-Lookup named entities, synonyms, or custom labels

-Build vocabularies or word embeddings

-Efficiently manage key-value NLP data structures

````python
word_pos = {

"Python": "NOUN",

"is": "AUX",

"awesome": "ADJ"

}

#OR
word_pos = dict(Python="NOUN", the ="AUX", awesome="ADJ")
````

````python
#Accessing & Updating

print(word_pos["Python"])
# 'NOUN'

word_pos["awesome"] = "ADV" # Update value

word_pos["learning"] = "VERB" # Add new key-value

#Removing Elements

word_pos.pop("the")
# Removes key 'is'

del word_pos["Python"]
# Also removes key 'Python'

word_pos.clear()
# Empties the dictionary
````
